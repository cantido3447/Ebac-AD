```python
import math
from functools import reduce

# EJERCICIO 1: Raíz Cuadrada (lambda vs trad)

# 1A: Sin lambda (función tradicional con def)
def raiz_cuadrada_tradicional(x):
    return math.sqrt(x)

numero = 16
print("Raíz cuadrada sin lambda:", raiz_cuadrada_tradicional(numero))


# 1B: Utilizando lambda
raiz_cuadrada_lambda = lambda x: math.sqrt(x)

print("Raíz cuadrada con lambda:", raiz_cuadrada_lambda(numero))


# EJERCICIO 2: Largo de palabras (map)

# 2A: Sin utilizar map
texto = "Python es un lenguaje poderoso para datos"
palabras = texto.split()

longitudes = []
for palabra in palabras:
    longitudes.append(len(palabra))

print("Longitudes sin map:", longitudes)


# 2B: Utilizando map
texto = "Python es un lenguaje poderoso para datos"
palabras = texto.split()

longitudes = list(map(len, palabras))

print("Longitudes con map:", longitudes)


# EJERCICIO 3: Producto de una lista (reduce)

# 3A: Sin utilizar reduce
numeros = [2, 3, 4, 5]

producto = 1
for n in numeros:
    producto *= n

print("Producto sin reduce:", producto)


# 3B: Utilizando reduce
numeros = [2, 3, 4, 5]

producto = reduce(lambda x, y: x * y, numeros)

print("Producto con reduce:", producto)


# EJERCICIO 4: Filtrar palabras con mayúsculas/números (filter)

# 4A: Sin utilizar filter
palabras = ["python", "Data1", "ciencia", "AI", "modelo3", "codigo"]

resultado = []
for palabra in palabras:
    if any(c.isupper() for c in palabra) or any(c.isdigit() for c in palabra):
        resultado.append(palabra)

print("Resultado sin filter:", resultado)


# 4B: Utilizando filter
palabras = ["python", "Data1", "ciencia", "AI", "modelo3", "codigo"]

resultado = list(filter(lambda palabra: any(c.isupper() or c.isdigit() for c in palabra), palabras))

print("Resultado con filter:", resultado)

```
