```python
print("--- 1. Manejo con try / except para archivo inexistente ---")
try:
    with open("test.txt", "r") as archivo:
        contenido = archivo.read()
except FileNotFoundError as e:
    print(f"Error detectado: El archivo 'test.txt' no existe en el directorio.")
    print(f"Detalle técnico: {e}")

print("\n--- 2. Estructura completa con try / except / else / finally ---")
try:
    print("Intentando abrir el archivo 'test.txt'...")
    archivo = open("test.txt", "r")
except FileNotFoundError:
    print("Excepción capturada: No se encontró el archivo solicitado.")
else:
    print("El archivo se leyó con éxito:")
    print(archivo.read())
    archivo.close()
finally:
    print("Proceso de lectura finalizado (bloque 'finally').")

print("\n--- 3. Depuración de operación matemática ---")
a = 10
b = 20
c = 30
d = 40

suma = a + b + c + d
print(f"Suma de los 4 números: {suma}")

try:
    resultado = suma / 0
except ZeroDivisionError as e:
    print(f"Error de división por cero capturado durante la depuración: {e}")
```
