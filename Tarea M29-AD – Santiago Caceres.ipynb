```python
import requests
from bs4 import BeautifulSoup
import json

print("=== EJERCICIO 1: Importación de sitio web con requests e impresión del contenido ===")
url_sitio = "https://www.wikipedia.org"
response = requests.get(url_sitio)
print(f"Código de Estado HTTP: {response.status_code}")
print("Primeros 400 caracteres del contenido HTML obtenido:\n")
print(response.text[:400])
print("\n" + "="*60 + "\n")

print("=== EJERCICIO 2: Uso de BeautifulSoup y requests (Parseo en sitio abierto) ===")
# Usamos una página web abierta y amigable para scraping (Quotes to Scrape / Toscrape)
url_scraping = "http://quotes.toscrape.com/"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/118.0.0.0 Safari/537.36"
}
response_scraping = requests.get(url_scraping, headers=headers)
print(f"Código de estado HTTP: {response_scraping.status_code}")

if response_scraping.status_code == 200:
    soup = BeautifulSoup(response_scraping.text, "html.parser")
    titulo = soup.title.string.strip() if soup.title else "Sin título"
    print(f"Título de la página web: {titulo}\n")
    print("Primeros elementos/frases detectadas en la página:")
    
    # Extraemos las citas (frases) de la página web abierta
    frases = soup.find_all('span', class_='text')
    contador = 1
    for tag in frases:
        texto = tag.get_text(strip=True)
        if texto:
            print(f"{contador}. {texto}")
            contador += 1
            if contador > 10:
                break
else:
    print(f"No se pudo acceder. Código de estado: {response_scraping.status_code}")

print("\n" + "="*60 + "\n")

print("=== EJERCICIO 3: Consumo de API sin autenticación y listado ordenado ===")
url_api = "http://universities.hipolabs.com/search?country=Argentina"
response_api = requests.get(url_api)
if response_api.status_code == 200:
    datos_json = response_api.json()
    lista_universidades = [item['name'] for item in datos_json]
    lista_ordenada = sorted(lista_universidades)
    print(f"Total de registros obtenidos de la API: {len(lista_ordenada)}\n")
    print("Listado ordenado alfabéticamente (Primeros 10 elementos):")
    for i, nombre in enumerate(lista_ordenada[:10], 1):
        print(f"{i}. {nombre}")
else:
    print(f"Error al conectar con la API. Código: {response_api.status_code}")
```
