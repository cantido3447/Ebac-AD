```python
import json
import pandas as pd

# ==========================================
# 1. ARCHIVO CSV (fifa_eda.csv)
# ==========================================

# Leer el archivo y guardarlo en una lista
df_csv = pd.read_csv('fifa_eda.csv')
lista_csv = df_csv.values.tolist()

# Obtener las 10 primeras filas
head_csv = df_csv.head(10)
print('--- 10 Primeras Filas (CSV) ---')
print(head_csv)

# Obtener las 10 últimas filas
tail_csv = df_csv.tail(10)
print('\n--- 10 Últimas Filas (CSV) ---')
print(tail_csv)

# Seleccionar 2-4 columnas y generar un nuevo archivo
columnas_csv = ['Name', 'Age', 'Nationality', 'Overall']
df_csv_subset = df_csv[columnas_csv]
df_csv_subset.to_csv('fifa_subset.csv', index=False)
print("\nArchivo 'fifa_subset.csv' generado correctamente.")


# ==========================================
# 2. ARCHIVO EXCEL (Amazon_top100_bestselling_books_2009to2021.xlsx)
# ==========================================

# Leer el archivo y guardarlo en una lista
df_excel = pd.read_excel('Amazon_top100_bestselling_books_2009to2021.xlsx')
lista_excel = df_excel.values.tolist()

# Obtener las 10 primeras filas
head_excel = df_excel.head(10)
print('\n--- 10 Primeras Filas (Excel) ---')
print(head_excel)

# Obtener las 10 últimas filas
tail_excel = df_excel.tail(10)
print('\n--- 10 Últimas Filas (Excel) ---')
print(tail_excel)

# Seleccionar 2-4 columnas y generar un nuevo archivo
columnas_excel = ['title', 'author', 'price', 'year']  # Columnas seleccionadas
df_excel_subset = df_excel[
    [col for col in columnas_excel if col in df_excel.columns]
]
df_excel_subset.to_excel('amazon_books_subset.xlsx', index=False)
print("\nArchivo 'amazon_books_subset.xlsx' generado correctamente.")


# ==========================================
# 3. ARCHIVO JSON (season5.json / Subtítulos GOT)
# ==========================================

# Leer el archivo JSON y cargarlo
with open('season5.json', 'r', encoding='utf-8') as f:
  datos_json = json.load(f)

# Aplanar la estructura del JSON para convertirlo a lista/DataFrame
filas_json = []
for episodio, subtitulos in datos_json.items():
  for line_id, texto in subtitulos.items():
    filas_json.append(
        {'Episodio': episodio, 'Line_ID': line_id, 'Texto_Subtitulo': texto}
    )

df_json = pd.DataFrame(filas_json)
lista_json = df_json.values.tolist()

# Obtener las 10 primeras filas
head_json = df_json.head(10)
print('\n--- 10 Primeras Filas (JSON) ---')
print(head_json)

# Obtener las 10 últimas filas
tail_json = df_json.tail(10)
print('\n--- 10 Últimas Filas (JSON) ---')
print(tail_json)

# Seleccionar 2-4 columnas y generar un nuevo archivo
df_json_subset = df_json[['Episodio', 'Texto_Subtitulo']]
df_json_subset.to_json(
    'got_subtitles_subset.json', orient='records', indent=4
)
print("\nArchivo 'got_subtitles_subset.json' generado correctamente.")
```
