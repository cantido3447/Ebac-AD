```python
import pandas as pd

print("--- 1. CARGANDO DATOS ---")
df_category = pd.read_csv('DIM_CATEGORY.csv')
df_product = pd.read_excel('DIM_PRODUCT.xlsx')
df_segment = pd.read_excel('DIM_SEGMENT.xlsx')
df_calendar = pd.read_excel('DIM_CALENDAR.xlsx')

print("DIM_CATEGORY cargado:", df_category.shape)
print("DIM_PRODUCT cargado:", df_product.shape)
print("DIM_SEGMENT cargado:", df_segment.shape)
print("DIM_CALENDAR cargado:", df_calendar.shape)

print("\n--- 2. REVISIÓN Y EXPLORACIÓN ---")
print("--- CATEGORY HEAD ---")
print(df_category.head())
print("--- PRODUCT INFO ---")
print(df_product.info())
print("--- SEGMENT DESCRIBE ---")
print(df_segment.describe(include='all'))

print("\n--- 3. LIMPIEZA DE DATOS ---")
df_category['CATEGORY'] = df_category['CATEGORY'].str.strip()
df_category = df_category.rename(columns={'ID_CATEGORY': 'CATEGORY_ID', 'CATEGORY': 'CATEGORY_NAME'})

print("Valores nulos en DIM_PRODUCT antes de limpieza:")
print(df_product.isnull().sum())
df_product['ITEM'] = df_product['ITEM'].fillna('UNKNOWN')
df_product['ATTR1'] = df_product['ATTR1'].fillna('NO DEFINIDO')
df_product['ATTR3'] = df_product['ATTR3'].fillna('NO DEFINIDO')

df_segment['ATTR3'] = df_segment['ATTR3'].fillna('NO DEFINIDO')

df_category = df_category.drop_duplicates()
df_product = df_product.drop_duplicates()
df_segment = df_segment.drop_duplicates()
df_calendar = df_calendar.drop_duplicates()

print("\n--- 4. UNIÓN Y CONSOLIDACIÓN ---")

df_merged = pd.merge(df_product, df_category, left_on='CATEGORY', right_on='CATEGORY_ID', how='left')


df_merged = pd.merge(df_merged, df_segment, on=['CATEGORY', 'FORMAT', 'ATTR1', 'ATTR2', 'ATTR3'], how='left', suffixes=('', '_seg'))

print("DataFrame consolidado final shape:", df_merged.shape)
print(df_merged.head())

output_filename = 'consolidated_dataset.xlsx'
df_merged.to_excel(output_filename, index=False)
print(f"\n--- 5. GUARDADO EXITOSO ---")
print(f"Dataset guardado en '{output_filename}'")
```
