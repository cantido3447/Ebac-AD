```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar el archivo fifa_eda.csv
df = pd.read_csv('fifa_eda.csv')

# 2. Mostrar las primeras 20 filas, las últimas 5 y un sample de 10
print("--- Primeras 20 filas ---")
display(df.head(20))

print("\n--- Últimas 5 filas ---")
display(df.tail(5))

print("\n--- Sample de 10 filas aleatorias ---")
display(df.sample(10))

# 3. Generar data estadística con .describe() y tipos de datos
print("\n--- Tipos de datos del dataset ---")
print(df.dtypes)

print("\n--- Data estadística (.describe()) ---")
display(df.describe())

# 4. Pasar a numéricas columnas si es necesario e incluirlas en .describe()
if 'Joined' in df.columns and not pd.api.types.is_numeric_dtype(df['Joined']):
    df['Joined'] = pd.to_numeric(df['Joined'], errors='coerce')

if 'Release Clause' in df.columns and not pd.api.types.is_numeric_dtype(df['Release Clause']):
    df['Release Clause'] = pd.to_numeric(df['Release Clause'], errors='coerce')

# Volvemos a aplicar describe para verificar las columnas numéricas actualizadas
display(df.describe())

# 5. Añadir una columna "Years Playing", que calcule el año actual menos la columna "Joined"
current_year = 2026
df['Years Playing'] = current_year - df['Joined']
print("\n--- Muestra con la nueva columna 'Years Playing' ---")
display(df[['Name', 'Joined', 'Years Playing']].head())

# 6. Buscar y mostrar a todos los jugadores de Mexico
mexico_players = df[df['Nationality'] == 'Mexico']
print(f"\n--- Jugadores de México (Total: {len(mexico_players)}) ---")
display(mexico_players)

# 7. Ordenar y mostrar los datos por la columna ReleaseClause (sueldo / cláusula de rescisión)
df_sorted = df.sort_values(by='Release Clause', ascending=False)
print("\n--- Datos ordenados por Release Clause (descendente) ---")
display(df_sorted[['Name', 'Club', 'Release Clause']].head(10))

# 8. Generar un nuevo dataset que contenga el año (joined) y el número de jugadores (groupby)
df_grouped = df.groupby('Joined').size().reset_index(name='Player_Count')
print("\n--- Número de jugadores por año de incorporación ---")
display(df_grouped)

# 9. (Opcional) Generar un gráfico que contenga, por año, el número de jugadores (Actualizado)
plt.figure(figsize=(10, 6))
sns.barplot(
    data=df_grouped, x='Joined', y='Player_Count', hue='Joined', palette='viridis', legend=False
)
plt.title('Número de jugadores por año de incorporación')
plt.xlabel('Año (Joined)')
plt.ylabel('Cantidad de Jugadores')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
