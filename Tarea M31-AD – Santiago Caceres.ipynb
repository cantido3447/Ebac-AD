```python
import pandas as pd

# 1. Cargar el dataset
df = pd.read_csv('fifa_eda.csv')

# Visualizar una muestra aleatoria de 20 líneas
print("--- MUESTRA DE 20 LÍNEAS ---")
print(df.sample(20))

# 2. Generar la media, mínimo y máximo de los jugadores en total
print("\n--- ESTADÍSTICAS GLOBALES ---")
estadisticas_globales = df[['Overall', 'Potential', 'Age', 'Value', 'Wage']].agg(['mean', 'min', 'max'])
print(estadisticas_globales)

# 3. Generar la media, mínimo y máximo por país (groupby)
print("\n--- ESTADÍSTICAS POR PAÍS ---")
estadisticas_por_pais = df.groupby('Nationality')[['Overall', 'Potential']].agg(['mean', 'min', 'max'])
print(estadisticas_por_pais.head(10))

# 4. Generar 2 jugadores ficticios y añadirlos al dataset
jugadores_ficticios = pd.DataFrame([
    {
        'ID': 999991, 'Name': 'Jugador Alfa', 'Age': 22, 'Nationality': 'Argentina', 
        'Overall': 88, 'Potential': 94, 'Club': 'Free Agent', 'Value': 50000.0, 
        'Wage': 100.0, 'Preferred Foot': 'Right', 'International Reputation': 2.0, 
        'Skill Moves': 4.0, 'Position': 'ST', 'Joined': 2026, 'Contract Valid Until': '2030-01-01', 
        'Height': 6.0, 'Weight': 165.0, 'Release Clause': 90000.0
    },
    {
        'ID': 999992, 'Name': 'Jugador Beta', 'Age': 25, 'Nationality': 'Brazil', 
        'Overall': 90, 'Potential': 92, 'Club': 'Free Agent', 'Value': 65000.0, 
        'Wage': 150.0, 'Preferred Foot': 'Left', 'International Reputation': 3.0, 
        'Skill Moves': 5.0, 'Position': 'CAM', 'Joined': 2026, 'Contract Valid Until': '2030-01-01', 
        'Height': 5.9, 'Weight': 155.0, 'Release Clause': 110000.0
    }
])

df_actualizado = pd.concat([df, jugadores_ficticios], ignore_index=True)

print("\n--- ÚLTIMAS FILAS CON JUGADORES FICTICIOS (.tail()) ---")
print(df_actualizado.tail(5)[['ID', 'Name', 'Nationality', 'Overall', 'Potential']])

print("\n--- NUEVAS ESTADÍSTICAS GLOBALES TRAS AÑADIR JUGADORES ---")
nuevas_estadisticas = df_actualizado[['Overall', 'Potential', 'Age', 'Value', 'Wage']].agg(['mean', 'min', 'max'])
print(nuevas_estadisticas)

# 5. Función lambda que añada la zona horaria y la capital del país de nacionalidad
info_paises = {
    'Argentina': {'Capital': 'Buenos Aires', 'Zona Horaria': 'UTC-3'},
    'Portugal': {'Capital': 'Lisboa', 'Zona Horaria': 'UTC+0'},
    'Brazil': {'Capital': 'Brasilia', 'Zona Horaria': 'UTC-3'},
    'Spain': {'Capital': 'Madrid', 'Zona Horaria': 'UTC+1'},
    'Belgium': {'Capital': 'Bruselas', 'Zona Horaria': 'UTC+1'}
}

df_actualizado['Capital'] = df_actualizado['Nationality'].apply(
    lambda x: info_paises.get(x, {}).get('Capital', 'Desconocida')
)

df_actualizado['Zona Horaria'] = df_actualizado['Nationality'].apply(
    lambda x: info_paises.get(x, {}).get('Zona Horaria', 'Desconocida')
)

# 6. Función lambda que en base al potencial añada "Candidato" o "Next Year" (> 85 son candidatos)
df_actualizado['Premio_Balon_Oro'] = df_actualizado['Potential'].apply(
    lambda p: 'Candidato' if p > 85 else 'Next Year'
)

print("\n--- MUESTRA FINAL CON COLUMNAS AÑADIDAS ---")
print(df_actualizado[['Name', 'Nationality', 'Capital', 'Zona Horaria', 'Potential', 'Premio_Balon_Oro']].head(10))
```
