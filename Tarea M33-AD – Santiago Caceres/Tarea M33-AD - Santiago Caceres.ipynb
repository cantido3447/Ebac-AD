```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# ==============================================================================
# 1. LECTURA, EXPLORACIÓN INICIAL Y VALIDACIÓN DE COLUMNAS NUMÉRICAS
# ==============================================================================
df = pd.read_csv("vgsales.csv")

print("=== HEAD ===")
display(df.head())

print("\n=== TAIL ===")
display(df.tail())

print("\n=== SAMPLE ===")
display(df.sample(5, random_state=42))

# Validación de columnas numéricas para asegurar compatibilidad
cols_ventas = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]
for col in cols_ventas + ["Year"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("\n=== DESCRIBE (COLUMNAS NUMÉRICAS VALIDADAS) ===")
display(df.describe())

# Dataset limpio con año válido como entero
df_clean = df.dropna(subset=["Year"]).copy()
df_clean["Year"] = df_clean["Year"].astype(int)

# ==============================================================================
# 2. TOTALES GENERALES
# ==============================================================================
total_registros = len(df)
total_juegos_unicos = df["Name"].nunique()
total_generos = df["Genre"].nunique()
generos_lista = df["Genre"].unique().tolist()
total_plataformas = df["Platform"].nunique()
plataformas_lista = df["Platform"].unique().tolist()

print(f"Total de registros (lanzamientos): {total_registros}")
print(f"1) Total de videojuegos únicos analizados: {total_juegos_unicos}")
print(f"2) Total de géneros analizados: {total_generos} -> {generos_lista}")
print(f"3) Total de plataformas analizadas: {total_plataformas} -> {plataformas_lista}")

# ==============================================================================
# 3. VENTAS DESGLOSADAS POR REGIÓN (NA, EU, JP, OTHER Y GLOBAL)
# ==============================================================================
# 3.1 Total de ventas por año
ventas_por_anio = df_clean.groupby("Year")[cols_ventas].sum().reset_index()
print("\n=== 1) TOTAL DE VENTAS POR AÑO (Muestra últimos 10 años) ===")
display(ventas_por_anio.tail(10))

# 3.2 Juegos más vendidos por año (agrupado por título y año, ordenado por Global_Sales)
juegos_por_anio = (
    df_clean.groupby(["Year", "Name"])[cols_ventas]
    .sum()
    .reset_index()
    .sort_values(["Year", "Global_Sales"], ascending=[True, False])
    .drop_duplicates(subset=["Year"])
)
print("\n=== 2) JUEGO MÁS VENDIDO POR AÑO (Muestra últimos 10 años) ===")
display(juegos_por_anio.tail(10))

# 3.3 Plataformas con más ingresos por año
plataformas_por_anio = (
    df_clean.groupby(["Year", "Platform"])[cols_ventas]
    .sum()
    .reset_index()
    .sort_values(["Year", "Global_Sales"], ascending=[True, False])
    .drop_duplicates(subset=["Year"])
)
print("\n=== 3) PLATAFORMA CON MÁS INGRESOS POR AÑO (Muestra últimos 10 años) ===")
display(plataformas_por_anio.tail(10))

# 3.4 Gráfico de comparación de ventas entre géneros por año
ventas_genero_anio = df_clean.pivot_table(
    index="Year", columns="Genre", values="Global_Sales", aggfunc="sum", fill_value=0
)

ax = ventas_genero_anio.plot(kind="line", linewidth=2, marker="o", markersize=4)
plt.title("Comparación de Ventas Globales por Género a lo largo de los Años", fontsize=14, fontweight="bold")
plt.xlabel("Año")
plt.ylabel("Ventas Globales (Millones)")
plt.legend(title="Género", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()

# ==============================================================================
# 4. INSIGHTS: 2 GRÁFICOS ADICIONALES INTERESANTES
# ==============================================================================
# Insight 1: Participación regional (%) por género (Preferencias culturales de mercado)
ventas_genero_reg = df_clean.groupby("Genre")[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]].sum()
pct_genero_reg = ventas_genero_reg.div(ventas_genero_reg.sum(axis=1), axis=0) * 100

pct_genero_reg.plot(kind="bar", stacked=True, colormap="viridis")
plt.title("Insight 1: Distribución Porcentual de Ventas Regionales por Género", fontsize=14, fontweight="bold")
plt.xlabel("Género")
plt.ylabel("Porcentaje de Ventas (%)")
plt.xticks(rotation=45)
plt.legend(["Norteamérica", "Europa", "Japón", "Otros"], bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()

# Insight 2: Top 10 Publishers con mayores ingresos globales y su desglose regional
top_publishers = (
    df_clean.groupby("Publisher")[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]]
    .sum()
    .sort_values("Global_Sales", ascending=False)
    .head(10)
    .drop(columns="Global_Sales")
)

top_publishers.plot(kind="barh", stacked=True, colormap="Set2")
plt.gca().invert_yaxis()
plt.title("Insight 2: Top 10 Publishers por Ingresos Globales y Región", fontsize=14, fontweight="bold")
plt.xlabel("Ventas (Millones)")
plt.ylabel("Publisher")
plt.legend(["Norteamérica", "Europa", "Japón", "Otros"], loc="lower right")
plt.tight_layout()
plt.show()

# ==============================================================================
# 5. ANÁLISIS DE CRECIMIENTO EN LOS ÚLTIMOS 4 AÑOS COMPLETOS (2013 - 2016)
# ==============================================================================
# Nota metodológica: 2017 (3 títulos) y 2020 (1 título) están incompletos en el dataset de Kaggle.
# Se toma la ventana completa de 4 años: 2013 a 2016.
df_4y = df_clean[df_clean["Year"].between(2013, 2016)].copy()

def analizar_crecimiento(dataframe, columna):
    piv = dataframe.pivot_table(
        index=columna, columns="Year", values="Global_Sales", aggfunc="sum", fill_value=0
    )
    piv["Crecimiento_Absoluto"] = piv[2016] - piv[2013]
    piv["Crecimiento_Pct"] = np.where(
        piv[2013] > 0, ((piv[2016] - piv[2013]) / piv[2013]) * 100, np.nan
    )
    return piv.sort_values("Crecimiento_Absoluto", ascending=False)

crec_genero = analizar_crecimiento(df_4y, "Genre")
crec_plataforma = analizar_crecimiento(df_4y, "Platform")
crec_juego = analizar_crecimiento(df_4y, "Name")

print("\n=== CRECIMIENTO POR GÉNERO (2013-2016) ===")
display(crec_genero)

print("\n=== CRECIMIENTO POR PLATAFORMA (2013-2016) ===")
display(crec_plataforma)

print("\n=== TOP 5 Y BOTTOM 5 CRECIMIENTO POR VIDEOJUEGO (2013-2016) ===")
display(pd.concat([crec_juego.head(5), crec_juego.tail(5)]))

# ==============================================================================
# 6. DOS ELEMENTOS ADICIONALES DE OPORTUNIDADES DE MERCADO (2013 - 2016)
# ==============================================================================
# Oportunidad 1: Rentabilidad promedio por título lanzado según Género (Océano Azul vs Saturación)
oportunidad_genero = (
    df_4y.groupby("Genre")
    .agg(
        Titulos_Lanzados=("Name", "count"),
        Ventas_Totales=("Global_Sales", "sum"),
        Venta_Promedio_Por_Titulo=("Global_Sales", "mean"),
    )
    .sort_values("Venta_Promedio_Por_Titulo", ascending=False)
)
print("\n=== OPORTUNIDAD DE MERCADO 1: EFICIENCIA POR GÉNERO (2013-2016) ===")
display(oportunidad_genero)

# Gráfico Oportunidad 1
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=oportunidad_genero.reset_index(),
    x="Titulos_Lanzados",
    y="Venta_Promedio_Por_Titulo",
    size="Ventas_Totales",
    hue="Genre",
    sizes=(100, 1000),
    legend=False,
)
for idx, row in oportunidad_genero.reset_index().iterrows():
    plt.text(row["Titulos_Lanzados"] + 8, row["Venta_Promedio_Por_Titulo"], row["Genre"], fontsize=9)
plt.title("Oportunidad 1: Saturación (Títulos) vs. Rentabilidad Promedio por Juego (2013-2016)", fontsize=13, fontweight="bold")
plt.xlabel("Cantidad de Títulos Lanzados")
plt.ylabel("Venta Promedio por Título (Millones)")
plt.tight_layout()
plt.show()

# Oportunidad 2: Rendimiento y cuota de mercado de Consolas de Nueva Generación (PS4, XOne, WiiU, 3DS, PC)
plataformas_activas = ["PS4", "XOne", "3DS", "WiiU", "PC"]
oportunidad_plat = (
    df_4y[df_4y["Platform"].isin(plataformas_activas)]
    .groupby("Platform")
    .agg(
        Titulos_Lanzados=("Name", "count"),
        Ventas_Totales=("Global_Sales", "sum"),
        Venta_Promedio_Por_Titulo=("Global_Sales", "mean"),
    )
    .sort_values("Ventas_Totales", ascending=False)
)
print("\n=== OPORTUNIDAD DE MERCADO 2: PLATAFORMAS ACTIVAS (2013-2016) ===")
display(oportunidad_plat)

oportunidad_plat["Venta_Promedio_Por_Titulo"].plot(kind="bar", color="#2b5c8f")
plt.title("Oportunidad 2: Ingreso Promedio por Título en Plataformas Activas (2013-2016)", fontsize=13, fontweight="bold")
plt.xlabel("Plataforma")
plt.ylabel("Venta Promedio por Título (Millones)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
